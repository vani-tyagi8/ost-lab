```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df = df.drop(columns=["Unnamed: 32"])
df.head()
df.isnull().sum()
```

```python
df['diagnosis'].value_counts()
```

```python
df['diagnosis'].value_counts().plot(kind='bar')
plt.title("Diagnosis Distribution (M vs B)")
plt.xlabel("Diagnosis")
plt.ylabel("Count")
plt.show()
```

```python
colors = df['diagnosis'].map({'M': 'red', 'B': 'blue'})
plt.scatter(df['radius_mean'], df['texture_mean'], c=colors)
plt.title("Radius Mean vs Texture Mean")
plt.xlabel("Radius Mean")
plt.ylabel("Texture Mean")

import matplotlib.patches as mpatches
red_patch = mpatches.Patch(color='red', label='Malignant (M)')
blue_patch = mpatches.Patch(color='blue', label='Benign (B)')
plt.legend(handles=[red_patch, blue_patch])

plt.show()
```
